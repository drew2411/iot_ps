```python
import os
import re

def rename_files_in_directory(directory_path):
    """
    Renames files in a given directory by stripping the text
    before a date in 'YYYY-MM-DD' format.

    Args:
        directory_path (str): The absolute path to the target directory.
    """
    # Regex to find a date in YYYY-MM-DD format.
    # \d{4} matches exactly four digits (the year).
    # \d{2} matches exactly two digits (the month/day).
    date_pattern = re.compile(r'\d{4}-\d{2}-\d{2}')

    print(f"Scanning directory: {directory_path}\n")

    if not os.path.isdir(directory_path):
        print(f"Error: Directory not found at '{directory_path}'")
        return

    try:
        # Get a list of all files in the directory
        for filename in os.listdir(directory_path):
            old_filepath = os.path.join(directory_path, filename)

            # Ensure we are only processing files, not subdirectories
            if os.path.isfile(old_filepath):
                # Search for the date pattern in the filename
                match = date_pattern.search(filename)

                if match:
                    # Get the starting position of the date
                    start_index = match.start()
                    
                    # Create the new filename starting from the date
                    new_filename = filename[start_index:]
                    new_filepath = os.path.join(directory_path, new_filename)

                    # Rename the file
                    print(f"Renaming: '{filename}'\n      to: '{new_filename}'")
                    os.rename(old_filepath, new_filepath)
                else:
                    # If no date is found, skip the file
                    print(f"Skipping: '{filename}' (no date pattern found)")
            else:
                print(f"Skipping: '{filename}' (this is a directory)")

        print("\nRenaming process complete.")

    except Exception as e:
        print(f"An error occurred: {e}")

# --- Main execution ---
if __name__ == "__main__":
    # Use a raw string (r"...") for Windows paths to handle backslashes correctly
    target_folder = r"C:\Users\nikhi\Downloads\VIT Downloads\big data\cat 2 nagraj"
    rename_files_in_directory(target_folder)

```
